```python
import os
import pandas as pd

# 현재 디렉토리: datasets_HSI
label_csv_path = "label/label.csv"
vs_path = "image/VS_image"
ts_path = "image/TS_image"

# ID 목록 로드
df = pd.read_csv(label_csv_path)
ids = df.iloc[:, 0].astype(str).tolist()

# 중복 여부 점검
both_exist = []
not_found = []
ok_ids = []

for obj_id in ids:
    in_vs = os.path.isdir(os.path.join(vs_path, obj_id))
    in_ts = os.path.isdir(os.path.join(ts_path, obj_id))
    
    if in_vs and in_ts:
        both_exist.append(obj_id)
    elif not in_vs and not in_ts:
        not_found.append(obj_id)
    else:
        ok_ids.append(obj_id)

# 결과 출력
print(f"✅ 한 곳에만 존재하는 정상 ID 수: {len(ok_ids)}")

if both_exist:
    print(f"⚠️ VS와 TS 모두에 존재하는 ID 수: {len(both_exist)}")
    for i in both_exist:
        print(f"  - {i}")
if not_found:
    print(f"❌ VS/TS 어디에도 없는 ID 수: {len(not_found)}")
    for i in not_found:
        print(f"  - {i}")
```

```python
import os
import pandas as pd

# 현재 디렉토리: datasets_HSI
label_csv_path = "label/label.csv"
vs_path = "image/VS_image"
ts_path = "image/TS_image"

# 1. label.csv에서 ID 목록 가져오기
df = pd.read_csv(label_csv_path)
label_ids = set(df.iloc[:, 0].astype(str))

# 2. VS/TS 폴더 안의 객체 폴더 이름 가져오기
vs_ids = set(os.listdir(vs_path)) if os.path.isdir(vs_path) else set()
ts_ids = set(os.listdir(ts_path)) if os.path.isdir(ts_path) else set()

# 3. 전체 폴더 ID 목록
all_image_ids = vs_ids.union(ts_ids)

# 4. label.csv에 없는 ID들 추출
unmatched_ids = sorted(all_image_ids - label_ids)

# 5. 결과 출력
if unmatched_ids:
    print(f"❌ label.csv에 없는 이미지 객체 수: {len(unmatched_ids)}")
    for obj_id in unmatched_ids:
        print(f"  - {obj_id}")
else:
    print("✅ VS/TS 이미지 폴더에 있는 모든 객체가 label.csv에도 존재합니다.")
```
