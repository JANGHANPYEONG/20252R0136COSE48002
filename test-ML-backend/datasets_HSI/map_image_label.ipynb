```python
import pandas as pd
import os

# CSV 파일 경로
csv_path = "label/label.csv"
df = pd.read_csv(csv_path)

# 이미지 루트 경로
image_root = "image"

# 대상 밴드
bands = [10, 60, 70, 100, 110]

# 경로 생성 함수
def find_image_path(obj_id, band):
    for image_type in ["VS_image", "TS_image"]:
        relative_path = f"{image_type}/{obj_id}/{obj_id}_P{band}.png"
        full_path = os.path.join(image_root, relative_path)
        if os.path.exists(full_path):
            return relative_path  # CSV에 넣을 상대 경로
    return None  # 둘 다 없을 경우

# path_XX 컬럼 생성
for band in bands:
    col_name = f"path_{band}"
    df[col_name] = df["ID"].apply(lambda obj_id: find_image_path(obj_id, band))

# 저장
df.to_csv(csv_path, index=False)
print("✅ 경로 삽입 완료! label.csv 파일에 path_XX 컬럼이 추가되었습니다.")
```
